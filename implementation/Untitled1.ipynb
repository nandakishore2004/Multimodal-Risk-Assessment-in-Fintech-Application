```python
import matplotlib.pyplot as plt

fig, ax = plt.subplots(figsize=(8, 10))
ax.set_xlim(0, 10)
ax.set_ylim(0, 12)
ax.axis('off')

# Function to draw a box
def box(x, y, text):
    ax.text(x, y, text,
            ha='center', va='center',
            fontsize=10,
            bbox=dict(boxstyle="round,pad=0.4",
                      fc="lightblue",
                      ec="black"))

# Function to draw arrow
def arrow(x1, y1, x2, y2):
    ax.annotate("",
                xy=(x2, y2),
                xytext=(x1, y1),
                arrowprops=dict(arrowstyle="->", lw=2))

# Boxes
box(5, 11, "START")

box(5, 10, "User Opens\nWeb Application")
arrow(5, 10.8, 5, 10.2)

box(5, 9, "Multimodal\nInput Form")
arrow(5, 9.8, 5, 9.2)

box(5, 8, "Image Upload\nVoice\nText\nTransaction")
arrow(5, 8.8, 5, 8.2)

box(5, 6.8, "Feature Extraction")
arrow(5, 7.8, 5, 7)

box(5, 5.6, "Risk Score Fusion")
arrow(5, 6.5, 5, 5.8)

box(5, 4.4, "Total Risk Score")
arrow(5, 5.3, 5, 4.6)

box(5, 3.2, "Decision")
arrow(5, 4.1, 5, 3.4)

box(2, 1.8, "Low Risk")
box(5, 1.8, "Medium Risk")
box(8, 1.8, "High Risk")

arrow(5, 3, 2, 2.1)
arrow(5, 3, 5, 2.1)
arrow(5, 3, 8, 2.1)

plt.savefig("flowchart.png", dpi=300, bbox_inches="tight")
plt.show()

print("Flowchart saved as flowchart.png")
```
