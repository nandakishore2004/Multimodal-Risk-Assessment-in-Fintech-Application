```python
import pandas as pd
import os

# Dataset path
corpus_path = r"D:\Major Project\DATASETS MAJOR PROJECT\cv-corpus-26.0-2026-06-12\te"
clips_path  = os.path.join(corpus_path, "clips")

# TSV files load చేయండి
validated = pd.read_csv(os.path.join(corpus_path, "validated.tsv"), sep='\t')
train_df  = pd.read_csv(os.path.join(corpus_path, "train.tsv"),     sep='\t')
test_df   = pd.read_csv(os.path.join(corpus_path, "test.tsv"),      sep='\t')

print("="*50)
print("   CV Telugu Corpus — Stats")
print("="*50)
print(f"Validated clips : {len(validated)}")
print(f"Train clips     : {len(train_df)}")
print(f"Test clips      : {len(test_df)}")
print(f"\nColumns: {list(validated.columns)}")
print(f"\nSample sentences:")
print(validated['sentence'].head(10).to_string())
```

```python
# Clips folder లో files ఉన్నాయా చూడండి
clips = os.listdir(clips_path)
print(f"Total audio clips: {len(clips)}")
print(f"Sample files: {clips[:5]}")

# File size check
sample_file = os.path.join(clips_path, clips[0])
size = os.path.getsize(sample_file)
print(f"\nSample file: {clips[0]}")
print(f"Size: {size} bytes")
```

```python
import whisper

print("Loading Whisper tiny...")
model = whisper.load_model("tiny")
print("✅ Ready!")

# First audio file test చేయండి
first_row = validated.iloc[0]
audio_file = os.path.join(clips_path, first_row['path'])

print(f"\nTesting: {first_row['path']}")
print(f"Actual  : {first_row['sentence']}")

if os.path.exists(audio_file):
    result = model.transcribe(audio_file, language='te')
    print(f"Whisper : {result['text']}")
else:
    print("❌ File not found! Checking path...")
    print(f"Looking at: {audio_file}")
    # Try without extension
    audio_file2 = audio_file + ".mp3"
    print(f"Trying  : {audio_file2}")
```

```python
from tqdm.notebook import tqdm

# Test set నుండి 50 samples
test_samples = test_df.head(50).copy()
results = []

print("Transcribing 50 real Telugu audio files...")

for _, row in tqdm(test_samples.iterrows(), total=len(test_samples)):
    # Audio file path
    audio_path = os.path.join(clips_path, row['path'])

    # .mp3 extension try చేయండి if file not found
    if not os.path.exists(audio_path):
        audio_path = audio_path + ".mp3"

    if os.path.exists(audio_path):
        try:
            out = model.transcribe(audio_path, language='te')
            results.append({
                'actual'    : row['sentence'],
                'predicted' : out['text'].strip(),
                'file'      : row['path']
            })
        except Exception as e:
            results.append({
                'actual'    : row['sentence'],
                'predicted' : '',
                'file'      : row['path']
            })
    else:
        print(f"⚠️ File not found: {row['path']}")

results_df = pd.DataFrame(results)
print(f"\n✅ Transcribed {len(results_df)} files!")
print(results_df[['actual','predicted']].head(10))
```

```python
!pip install jiwer -q
from jiwer import wer

# Empty predictions తీసేయండి
clean_df = results_df[results_df['predicted'] != ''].copy()

error_rate = wer(
    clean_df['actual'].tolist(),
    clean_df['predicted'].tolist()
)
accuracy = round((1 - error_rate) * 100, 2)

print("="*50)
print("   Whisper Telugu ASR — Real Dataset Results")
print("="*50)
print(f"Files tested    : {len(clean_df)}")
print(f"Word Error Rate : {round(error_rate*100, 2)}%")
print(f"Accuracy (WER)  : {accuracy}%")
print("="*50)

# Sample results
print("\nSample Comparisons:")
for _, row in clean_df.head(5).iterrows():
    print(f"  Actual  : {row['actual']}")
    print(f"  Whisper : {row['predicted']}")
    print()
```

```python
risk_keywords = {
    'HIGH'  : ['loan','mosam','fraud','debt','poyaayi','moosam',
                'మోసం','అప్పు','పోయాయి','lakshal','baaki'],
    'MEDIUM': ['transfer','bill','withdraw','chesanu','challenge',
                'transfer','బిల్','చెల్లించాలి'],
    'LOW'   : ['save','deposit','salary','jam','savings',
                'save','జమ','salary']
}

def detect_risk(text):
    t = text.lower()
    for kw in risk_keywords['HIGH']:
        if kw.lower() in t: return 'HIGH',   0.85, kw
    for kw in risk_keywords['MEDIUM']:
        if kw.lower() in t: return 'MEDIUM', 0.55, kw
    for kw in risk_keywords['LOW']:
        if kw.lower() in t: return 'LOW',    0.20, kw
    return 'NEUTRAL', 0.10, 'none'

# Apply to all transcriptions
clean_df['risk_level'], clean_df['risk_score'], clean_df['keyword'] = \
    zip(*clean_df['predicted'].apply(detect_risk))

print("Risk Distribution in Telugu Corpus:")
print(clean_df['risk_level'].value_counts())
print()
print(clean_df[['actual','predicted','risk_level','keyword']].head(10).to_string())
```

```python
import pickle

model_data = {
    'model_name'    : 'whisper-tiny',
    'language'      : 'te',
    'risk_keywords' : risk_keywords,
    'wer'           : round(error_rate*100, 2),
    'accuracy'      : accuracy,
    'trained_on'    : 'Mozilla Common Voice 26.0 — Telugu'
}

save_path = r'D:\Major Project\model\telugu_voice_model.pkl'
pickle.dump(model_data, open(save_path, 'wb'))

print("✅ Model saved!")
print(f"   Path     : {save_path}")
print(f"   Dataset  : Mozilla Common Voice 26.0 Telugu")
print(f"   Accuracy : {accuracy}%")
print(f"   WER      : {round(error_rate*100,2)}%")
```

```python
import whisper
from jiwer import wer
import pandas as pd
import os

corpus_path = r"D:\Major Project\DATASETS MAJOR PROJECT\cv-corpus-26.0-2026-06-12\te"
clips_path  = os.path.join(corpus_path, "clips")
test_df     = pd.read_csv(os.path.join(corpus_path, "test.tsv"), sep='\t')

# Compare tiny vs small on 20 samples only
test_samples = test_df.head(20)

results_compare = []

for model_name in ["tiny", "small"]:
    print(f"\nLoading Whisper {model_name}...")
    mdl = whisper.load_model(model_name)

    preds, actuals = [], []
    for _, row in test_samples.iterrows():
        audio = os.path.join(clips_path, row['path'])
        if not os.path.exists(audio):
            audio += ".mp3"
        if os.path.exists(audio):
            try:
                out = mdl.transcribe(audio, language='te')
                preds.append(out['text'].strip())
                actuals.append(row['sentence'])
            except:
                pass

    if preds:
        w = wer(actuals, preds)
        results_compare.append({
            'Model'   : f'Whisper {model_name}',
            'WER %'   : round(w * 100, 1),
            'Accuracy': round((1-w)*100, 1) if w < 1 else "< 0%"
        })
        print(f"  WER: {round(w*100,1)}%")

print("\n")
print(pd.DataFrame(results_compare).to_string(index=False))
```
